# Computer Vision
Computer Vision is when we teach computers to "see" and understand images or videos — like how humans use their eyes and brain.

### Examples:

Recognizing faces in photos

Detecting objects in a picture (like cars, cats, or people)

Reading text from an image (OCR)

Checking if an X-ray has signs of disease

## 1. Reading Image

## 2. Showing Image

```python
# First importing the library for showing the image
from google.colab.patches import cv2_imshow
```

```python
# Showing the image
cv2_imshow(image)
```

## 3. Channels

```python
# Making a copy to make changes on the image
image1 = image.copy()
```

```python
# Instead of RGB we use BGR in libraries -  BLUE, GREEN, RED
# Chanelling the blue one as its on 0 index and putting its intensity value as 0
image1[:, :, 0] = 0
cv2_imshow(image1)
```

```python
# Chanelling the green one as its on 1 index and putting its intensity value as 0
image1[:, :, 1] = 0
cv2_imshow(image1)

# Now we are left with only red channel
```

## 4. Converting Image

```python
cv2_imshow(image)
```

```python
image.shape
```

```python
image2 = image.copy()
```

```python
# Converting image to grayscale
# Gray scale has only one channel
gray = cv2.cvtColor(image2,cv2.COLOR_BGR2GRAY)
cv2_imshow(gray)
```

```python
# As grayscale has only one channel, so it's not shown
gray.shape
```

## 5. Flipping Image

```python
# +1 flips the image horizontally, -1 flips the image both, 0 flips it vertially
hflip = cv2.flip(gray, -1)
cv2_imshow(hflip)

# Using -1 flipped it horizontally as well as vertically
```

## 6. Cropping Image
Rows (top to bottom → height)

Columns (left to right → width)

Channels (color → Red, Green, Blue)

```python
image.shape
```

```python
# Similar to slicing
# getting height from 4 to 299, width from 100 to 799 and all the available channels
cropped = image[4:300, 100:800, :]
cv2_imshow(cropped)
```

## 7. Blurring An Image

```python
# ksize=(7, 7) is the size of the blur window (kernel). It looks at a 7×7 square of pixels to compute the blur.
# sigmaX=0, sigmaY=0. These control how strong the blur is in X (horizontal) and Y (vertical) directions.
blurr = cv2.GaussianBlur(image, ksize =(15, 15), sigmaX = 0, sigmaY = 0)
cv2_imshow(blurr)
```

## 8. Resizing Image

```python
image.shape
```

```python
# Resizing our image
resized = cv2.resize(image, dsize = (700, 200))
cv2_imshow(resized)
```

## 9. Rescaling Image

```python
# Rescaling the image
# factor = 0.7 - It scales the image down to 70% of the original size.
factor = 0.7

# image.shape[0] → number of rows (height), image.shape[1] → number of columns (width)
height = int(image.shape[0] * factor)
width = int(image.shape[1] * factor)

# Using resize to rescale it
rescaled = cv2.resize(image, dsize = (width, height))
cv2_imshow(rescaled)
```

## 10. Circle On Image

```python
# color = (0, 34, 255) - its, 0 intensity for Blue, 155 for green, 155 for red
# center=(500, 600) - The center of the circle is (x=500, y=600)
# radius=300	How big the circle is.
# color=(0, 155, 155)	The circle's color (BGR)
# thickness=3	How thick the circle line is.
circleImg = cv2.circle(image, center = (200, 300), radius = 100, thickness = 2, color = (0, 155, 155))
cv2_imshow(circleImg)
```

## 11. Rectangle On Image

```python
# Making a rectangle on our image
# pt1=(3, 3) - The starting point of the line (x=3, y=3)
# pt2=(600, 610) - The ending point of the line (x=600, y=610)
recImage = cv2.rectangle(image, pt1 =(3, 3,), pt2 = (200, 210), color = (0, 155, 155), thickness = 2)
cv2_imshow(recImage)
```

## 12. Line On Image

```python
# Drawing a line on the image
lineImg = cv2.line(image, pt1= (1, 2), pt2 = (300, 400), color = (0, 155, 155), thickness = 2)
cv2_imshow(lineImg)
```

## 13. Text On Image

```python
# Writing text on the Image
# org=(200, 500) - Position of the text on the image: (x=200, y=500)
# fontFace=FONT_ITALIC - The style of the text (basic and clean font)
# fontScale=3 - The size of the text (1 = normal size)
textImg = cv2.putText(image, text = 'Hello! Noor here!',
         org = (150, 440),
         fontFace = cv2.FONT_ITALIC,
         fontScale = 1,
         color = (140, 155, 50), thickness = 2)

cv2_imshow(textImg)
```
